# 01 — EDA
Exploratory look at the cleaned sales data. Run `src/data_generation.py` and `src/preprocessing.py` first.

```
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/sales_clean.csv', parse_dates=['date'])
daily = pd.read_csv('../data/processed/daily_totals.csv', parse_dates=['date'])
df.head()
```

## Overall shape and summary stats

```
print(df.shape)
df.describe()
```

## Daily units sold over time

```
plt.figure(figsize=(12,4))
plt.plot(daily['date'], daily['units_sold'])
plt.title('Daily Units Sold')
plt.xlabel('Date'); plt.ylabel('Units Sold')
plt.show()
```

## Sales by store and category

```
df.groupby('store')['units_sold'].sum().plot(kind='bar', title='Total Units Sold by Store')
plt.show()
```

```
df.groupby('category')['units_sold'].sum().plot(kind='bar', title='Total Units Sold by Category')
plt.show()
```

## Weekly seasonality check

```
daily['day_of_week'] = daily['date'].dt.day_name()
daily.groupby('day_of_week')['units_sold'].mean().reindex(
    ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
).plot(kind='bar', title='Avg Units Sold by Day of Week')
plt.show()
```
